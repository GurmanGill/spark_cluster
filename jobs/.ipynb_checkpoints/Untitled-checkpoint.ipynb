```python
import sys
import pyspark
import py4j

print(sys.executable)
print("PySpark:", pyspark.__version__)
print("Py4J:", py4j.__version__)
```

```python
from pyspark.sql import SparkSession
from pyspark.sql import functions as F

spark = (
    SparkSession.builder
    .appName("Stages-Tasks-Demo")
    .master("spark://spark-master:7077")
    .config("spark.executor.memory", "512m")
    .config("spark.executor.cores", "1")
    .config("spark.sql.shuffle.partitions", "8")
    .getOrCreate()
)

spark.sparkContext.setLogLevel("ERROR")


# --------------------------------------------------
# 1. Create a reasonably large distributed dataset
# --------------------------------------------------

df = (
    spark.range(0, 5_000_000, 1, numPartitions=8)
    .withColumn("group_id", (F.col("id") % 1000).cast("int"))
    .withColumn("value", (F.col("id") * 10) % 5000)
)

print("Initial partitions:", df.rdd.getNumPartitions())


# --------------------------------------------------
# 2. Force a shuffle using repartition
# --------------------------------------------------

repartitioned = df.repartition(16, "group_id")

print("After repartition:", repartitioned.rdd.getNumPartitions())


# --------------------------------------------------
# 3. Create a second dataset for a join
# --------------------------------------------------

lookup = (
    spark.range(0, 1000)
    .withColumnRenamed("id", "group_id")
    .withColumn("category", F.concat(F.lit("category_"), F.col("group_id")))
)


# --------------------------------------------------
# 4. Join
#
# Disable broadcast so Spark has to do a real shuffle join.
# --------------------------------------------------

spark.conf.set("spark.sql.autoBroadcastJoinThreshold", -1)

joined = repartitioned.join(
    lookup,
    on="group_id",
    how="inner"
)


# --------------------------------------------------
# 5. Aggregation
#
# groupBy causes another shuffle.
# --------------------------------------------------

aggregated = (
    joined
    .groupBy("category")
    .agg(
        F.count("*").alias("row_count"),
        F.sum("value").alias("total_value"),
        F.avg("value").alias("avg_value")
    )
)


# --------------------------------------------------
# 6. Sort
#
# Global orderBy causes another shuffle.
# --------------------------------------------------

result = aggregated.orderBy(F.desc("total_value"))


# --------------------------------------------------
# 7. ACTION
#
# Nothing substantial runs until this point because
# DataFrame transformations are lazy.
# --------------------------------------------------

result.show(50, truncate=False)
```
